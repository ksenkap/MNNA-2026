```python
"""ЗАДАНИЕ 3.2 ОЧИСТКА ДАННЫХ"""
import re
import os
import shutil

# 1. Читаем наш локальный сырой файл
input_path = "/content/raw_data.txt"
output_path = "/content/cleaned_dataset.txt"

print("Очистка данных (3.2)...")

with open(input_path, 'r', encoding='utf-8') as f:
    raw_text = f.read()

# 2. Удаляем пустые строки и нормализуем пробелы
lines = [line.strip() for line in raw_text.split('\n') if line.strip()]

# 3. Удаляем неизвестные языки и символы (оставляем кириллицу, латиницу, цифры и базовую пунктуацию)
allowed_chars = r'[^a-zA-Zа-яА-ЯёЁ0-9 .,!?;:()\-«»"\'\n]'
filtered_lines = []
for line in lines:
    # Нормализуем строку (убираем лишние пробелы внутри)
    line = " ".join(line.split())
    # Убираем все недопустимые символы
    line = re.sub(allowed_chars, '', line)
    # Оставляем только строки, которые длиннее 20 символов (отсекаем мусор и огрызки)
    if len(line) > 20:
        filtered_lines.append(line)

clean_text = '\n'.join(filtered_lines)

# 4. Разбиваем слишком длинные объекты на куски (чанки) по ~500 слов
words = clean_text.split()
chunks = []
chunk_size = 500

for i in range(0, len(words), chunk_size):
    chunk = ' '.join(words[i:i + chunk_size])
    if len(chunk) > 100: # Отбрасываем слишком короткие "хвосты"
        chunks.append(chunk)

# 5. Сохраняем локально очищенный датасет
with open(output_path, 'w', encoding='utf-8') as f:
    f.write('\n\n'.join(chunks))

print(f"Очистка завершена")
print(f"Получено объектов (чанков): {len(chunks)}")
print(f"Размер очищенного файла: {os.path.getsize(output_path) / (1024*1024):.2f} MB")

# 6. Переносим итоговый очищенный файл на Google Drive
drive_path = "/content/drive/MyDrive/CommonCrawl/cleaned_dataset.txt"
shutil.copy(output_path, drive_path)
print(f"Файл успешно сохранен на Google Drive: {drive_path}")
```

```python
"""ЗАДАНИЕ 3.3 УДАЛЕНИЕ ДУБЛИКАТОВ И ЭНТРОПИЯ"""
!pip install transformers torch -q

import torch
import numpy as np
from transformers import GPT2LMHeadModel, GPT2TokenizerFast
from tqdm import tqdm
import os

# --- 1. Загрузка данных ---
input_path = "/content/drive/MyDrive/CommonCrawl/cleaned_dataset.txt"
output_path = "/content/drive/MyDrive/CommonCrawl/filtered_dataset.txt"

print("Загрузка очищенного датасета...")
with open(input_path, 'r', encoding='utf-8') as f:
    text = f.read()

chunks = [chunk.strip() for chunk in text.split('\n\n') if chunk.strip()]
print(f"Исходное количество объектов: {len(chunks)}")

# --- 2. Инициализация GPT-2 ---
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Используемое устройство: {device}")

model_id = "gpt2"
tokenizer = GPT2TokenizerFast.from_pretrained(model_id)
# Устанавливаем pad_token
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

model = GPT2LMHeadModel.from_pretrained(model_id).to(device)
model.eval()

# --- 3. Функция подсчета энтропии (по батчам) ---
def calculate_entropy_batched(texts, batch_size=2):
    """
    Считает суммарную энтропию H(D) для каждого текста, обрабатывая их батчами.
    """
    entropies = []

    # Разбиваем список на батчи
    for i in tqdm(range(0, len(texts), batch_size), desc="Подсчет энтропии"):
        batch_texts = texts[i:i+batch_size]

        # Токенизируем батч
        encodings = tokenizer(batch_texts, return_tensors="pt", padding=True, truncation=True, max_length=1024).to(device)

        with torch.no_grad():
            # Считаем loss для каждого элемента в батче
            # Для этого нам нужно передать labels, но loss усредняется по батчу.
            # Чтобы получить loss для каждого объекта, мы посчитаем его вручную через logits.
            outputs = model(**encodings)
            logits = outputs.logits # (batch_size, seq_len, vocab_size)

            # Сдвигаем logits и labels для подсчета кросс-энтропии
            shift_logits = logits[..., :-1, :].contiguous()
            shift_labels = encodings["input_ids"][..., 1:].contiguous()

            # Считаем кросс-энтропию для каждого токена
            loss_fct = torch.nn.CrossEntropyLoss(reduction='none')
            losses = loss_fct(shift_logits.view(-1, shift_logits.size(-1)), shift_labels.view(-1))

            # Возвращаем форму (batch_size, seq_len-1)
            losses = losses.view(shift_labels.size(0), -1)

            # Учитываем маску внимания (padding токены не должны влиять)
            # Сдвигаем маску
            mask = encodings["attention_mask"][..., 1:].contiguous()

            # Считаем сумму энтропии (H(D) = -sum(log p)) для каждого объекта
            for j in range(len(batch_texts)):
                # Берем только реальные токены (не padding)
                valid_losses = losses[j][mask[j] == 1]
                # Суммарная энтропия для объекта
                total_entropy = valid_losses.sum().item()
                entropies.append(total_entropy)

    return entropies

# --- 4. Запуск подсчета для ВСЕХ объектов ---
print(f"Подсчет энтропии для всех {len(chunks)} объектов...")
# Если GPU не хватает, уменьшите batch_size до 2 или 1
entropies = calculate_entropy_batched(chunks, batch_size=2)

# --- 5. Анализ и Фильтрация ---
# 5.1 Удаление полных дубликатов (сначала удаляем, чтобы не тратить время на них)
unique_chunks = list(dict.fromkeys(chunks))
print(f"Удалено дубликатов: {len(chunks) - len(unique_chunks)}")

# Важно: энтропии мы считали для исходного списка chunks.
# Если дубликатов много, нужно пересчитать энтропию для уникальных,
# либо сопоставить их. Для простоты будем считать, что энтропии соответствуют
# исходному списку, и отфильтруем по порогам.

entropies_np = np.array(entropies)

# Определяем пороги (например, 10-й и 90-й процентиль)
low_threshold = np.percentile(entropies_np, 10)
high_threshold = np.percentile(entropies_np, 90)

print(f"Порог низкой энтропии (10%): {low_threshold:.2f}")
print(f"Порог высокой энтропии (90%): {high_threshold:.2f}")

# Фильтруем объекты, которые НЕ являются дубликатами и попадают в пороги
seen = set()
filtered_chunks = []
for chunk, ent in zip(chunks, entropies):
    if chunk in seen:
        continue
    if low_threshold <= ent <= high_threshold:
        filtered_chunks.append(chunk)
        seen.add(chunk)

# 5.3 Оценка информационной плотности всего датасета
total_entropy = sum(entropies)
# Считаем общее количество токенов во всех объектах
total_tokens = sum([len(tokenizer.encode(c)) for c in chunks])
rho_info = total_entropy / total_tokens

print(f"\nИнформационная плотность датасета (rho_info): {rho_info:.4f} бит/токен")
print(f"Оставлено объектов после фильтрации: {len(filtered_chunks)} из {len(chunks)}")

# --- 6. Сохранение результата ---
with open(output_path, 'w', encoding='utf-8') as f:
    f.write('\n\n'.join(filtered_chunks))

print(f"Отфильтрованный датасет сохранен: {output_path}")
```

```python
"""ПОДСЧЕТ ЧАСТИЧНЫХ ДУБЛИКАТОВ"""
#Здесь ищутся повторяющиеся n-граммы (фразы из 10 слов) внутри всего датасета.

from collections import Counter
import re

# 1. Загружаем отфильтрованный датасет
input_path = "/content/drive/MyDrive/CommonCrawl/filtered_dataset.txt"

with open(input_path, 'r', encoding='utf-8') as f:
    text = f.read()

chunks = [c.strip() for c in text.split('\n\n') if c.strip()]
print(f"Объектов для анализа: {len(chunks)}")

# 2. Собираем n-граммы (фразы из N слов)
N = 10  # Длина фразы. 10 слов — хороший баланс между точностью и скоростью
ngrams = []
for chunk in chunks:
    words = chunk.split()
    for i in range(len(words) - N + 1):
        ngrams.append(' '.join(words[i:i+N]))

print(f"Всего n-грамм: {len(ngrams)}")

# 3. Считаем частоты
ngram_counts = Counter(ngrams)

# 4. Анализ
total_ngrams = len(ngrams)
unique_ngrams = len(ngram_counts)

# Дублирующая плотность (rho_dub)
# Это отношение общего числа n-грамм к числу уникальных
rho_dub = total_ngrams / unique_ngrams if unique_ngrams > 0 else 0

# Количество n-грамм, которые встретились больше 1 раза
duplicated_ngrams = sum(count for count in ngram_counts.values() if count > 1)

print(f"\n--- Результаты анализа дубликатов (N={N}) ---")
print(f"Всего n-грамм: {total_ngrams}")
print(f"Уникальных n-грамм: {unique_ngrams}")
print(f"Повторяющихся n-грамм: {duplicated_ngrams}")
print(f"Дублирующая плотность (rho_dub): {rho_dub:.4f}")

# 6. Оценка: много это или мало?
if rho_dub < 1.05:
    print("\nДатасет очень чистый. Дубликатов практически нет.")
elif rho_dub < 1.2:
    print("\nУмеренное количество дубликатов. Это нормально для Common Crawl.")
else:
    print("\nМного дубликатов! Рекомендуется дополнительная очистка.")
```

```python
"""Необязательный блок поска и удаления почти-дубликатов, т.е. здесь ищутся пары объектов, которые очень похожи друг на друга.
Мы берем два объекта (например, две статьи) и сравниваем их множества n-грамм. Если 80%+ n-грамм совпадают — это почти-дубликаты."""

from collections import Counter
import re

# Загружаем датасет
with open("/content/drive/MyDrive/CommonCrawl/filtered_dataset.txt", 'r', encoding='utf-8') as f:
    text = f.read()
chunks = [c.strip() for c in text.split('\n\n') if c.strip()]

N = 10 # Длина n-грамм

# Функция для получения множества n-грамм объекта
def get_ngrams(text, n):
    words = text.split()
    return set(' '.join(words[i:i+n]) for i in range(len(words)-n+1))

# Считаем n-граммы для каждого объекта
chunk_ngrams = [get_ngrams(c, N) for c in chunks]

# Ищем почти-дубликаты (схожесть > 0.8)
to_remove = set()
threshold = 0.8

print("Поиск почти-дубликатов...")
for i in range(len(chunks)):
    if i in to_remove:
        continue
    for j in range(i+1, len(chunks)):
        if j in to_remove:
            continue

        # Jaccard similarity
        intersection = len(chunk_ngrams[i] & chunk_ngrams[j])
        union = len(chunk_ngrams[i] | chunk_ngrams[j])

        if union == 0:
            continue

        similarity = intersection / union

        if similarity > threshold:
            # Оставляем более длинный объект, удаляем короткий
            if len(chunks[i]) >= len(chunks[j]):
                to_remove.add(j)
            else:
                to_remove.add(i)
                break

# Формируем очищенный список
cleaned_chunks = [chunks[i] for i in range(len(chunks)) if i not in to_remove]

print(f"Найдено 80%-дубликатов: {len(to_remove)}")
print(f"Осталось объектов: {len(cleaned_chunks)}")

# Сохраняем
with open("/content/drive/MyDrive/CommonCrawl/deduplicated_dataset.txt", 'w', encoding='utf-8') as f:
    f.write('\n\n'.join(cleaned_chunks))
```

```python
"""ДОПОЛНИТЕЛЬНАЯ ФИЛЬТРАЦИЯ ДЛЯ ТОКЕНИЗАЦИИ ПО СЛОВАМ""" #Возможно, правильнее будет сделать и до символьной токенизации
import re
from collections import Counter

# Загружаем датасет
with open("/content/drive/MyDrive/CommonCrawl/filtered_dataset.txt", 'r', encoding='utf-8') as f:
    text = f.read()
chunks = [c.strip() for c in text.split('\n\n') if c.strip()]

print(f"Объектов до очистки: {len(chunks)}")

def is_good_text(chunk):
    """Эвристика: является ли объект связным текстом?"""
    # 1. Слишком короткий?
    if len(chunk) < 100:
        return False

    words = chunk.split()
    if len(words) < 20:
        return False

    # 2. Слишком много цифр?
    digits = sum(1 for c in chunk if c.isdigit())
    if digits / len(chunk) > 0.3: # >30% цифр
        return False

    # 3. Слишком много знаков препинания?
    punctuation = sum(1 for c in chunk if c in '.,!?;:()[]{}"\'')
    if punctuation / len(chunk) > 0.3:
        return False

    # 4. Слишком много повторяющихся знаков (!!!!, ?????)
    if re.search(r'([!?.,])\1{3,}', chunk): # 4+ одинаковых знака подряд
        return False

    # 5. Слишком высокая доля уникальных слов (признак списка/технической информации)
    unique_words = len(set(w.lower() for w in words))
    if unique_words / len(words) > 0.9: # >90% слов уникальны
        return False

    # 6. Слишком много строк, начинающихся с дефиса или цифры (списки)
    lines = chunk.split('\n')
    list_lines = sum(1 for line in lines if re.match(r'^\s*[-*\d]', line))
    if len(lines) > 0 and list_lines / len(lines) > 0.5:
        return False

    return True

# Фильтруем
clean_chunks = [c for c in chunks if is_good_text(c)]

print(f"Объектов после очистки: {len(clean_chunks)}")
print(f"Удалено 'мусора': {len(chunks) - len(clean_chunks)}")

# Сохраняем
with open("/content/drive/MyDrive/CommonCrawl/filtered_dataset_v2.txt", 'w', encoding='utf-8') as f:
    f.write('\n\n'.join(clean_chunks))
```

```python
"""BPE-ТОКЕНИЗАЦИЯ ВРУЧНУЮ"""
import random
from collections import Counter, defaultdict

# 1. Загружаем датасет
input_path = "/content/drive/MyDrive/CommonCrawl/filtered_dataset_v2.txt"

with open(input_path, 'r', encoding='utf-8') as f:
    text = f.read()

chunks = [c.strip() for c in text.split('\n\n') if c.strip()]
print(f"Загружено объектов: {len(chunks)}")

# 2. Берем подвыборку для обучения (BPE с нуля медленный на больших данных)
# Возьмем первые 200 объектов для демонстрации
train_chunks = chunks[:200]
print(f"Обучение BPE на {len(train_chunks)} объектах...")

# 3. Подготовка: разбиваем текст на слова и представляем каждое слово как список символов + маркер конца слова
# Например, "кот" -> ['к', 'о', 'т', '</w>']
def prepare_corpus(texts):
    word_freqs = Counter()
    for text in texts:
        for word in text.lower().split():
            # Добавляем маркер конца слова, чтобы BPE знал, где заканчивается слово
            word_freqs[tuple(word) + ('</w>',)] += 1
    return word_freqs

word_freqs = prepare_corpus(train_chunks)
print(f"Уникальных слов в подвыборке: {len(word_freqs)}")

# 4. Функция для получения всех пар символов
def get_stats(vocab):
    pairs = Counter()
    for word, freq in vocab.items():
        for i in range(len(word) - 1):
            pairs[(word[i], word[i+1])] += freq
    return pairs

# 5. Функция для объединения пары во всех словах
def merge_vocab(pair, vocab):
    new_vocab = {}
    bigram = ' '.join(pair)
    replacement = ''.join(pair)
    for word, freq in vocab.items():
        # Проходим по слову и заменяем пару на объединенный токен
        new_word = []
        i = 0
        while i < len(word):
            if i < len(word) - 1 and word[i] == pair[0] and word[i+1] == pair[1]:
                new_word.append(replacement)
                i += 2
            else:
                new_word.append(word[i])
                i += 1
        new_vocab[tuple(new_word)] = freq
    return new_vocab

# 6. Основной цикл BPE
num_merges = 500  # Сколько раз объединять пары. Чем больше — тем крупнее токены.
vocab = dict(word_freqs)
merges = []  # Список правил объединения (для декодирования)

print(f"\n--- Обучение BPE ({num_merges} операций объединения) ---")
for i in range(num_merges):
    pairs = get_stats(vocab)
    if not pairs:
        break
    best_pair = max(pairs, key=pairs.get)
    vocab = merge_vocab(best_pair, vocab)
    merges.append(best_pair)
    if (i + 1) % 100 == 0:
        print(f"  Итерация {i+1}/{num_merges}: объединяем {best_pair} (частота {pairs[best_pair]})")

# 7. Формируем итоговый словарь токенов
vocab_tokens = set()
for word in vocab.keys():
    for token in word:
        vocab_tokens.add(token)

print(f"\n--- Результаты чистой BPE-реализации ---")
print(f"Размер словаря: {len(vocab_tokens)}")
print(f"Примеры токенов: {list(vocab_tokens)[:20]}")

# 8. Функция токенизации с использованием выученных merges
def bpe_tokenize(word, merges):
    """Применяет выученные правила к новому слову"""
    tokens = list(word.lower()) + ['</w>']
    for pair in merges:
        new_tokens = []
        i = 0
        while i < len(tokens):
            if i < len(tokens) - 1 and tokens[i] == pair[0] and tokens[i+1] == pair[1]:
                new_tokens.append(''.join(pair))
                i += 2
            else:
                new_tokens.append(tokens[i])
                i += 1
        tokens = new_tokens
    return tokens

# 9. Тестируем на случайном объекте
random_chunk = random.choice(chunks)
print(f"\nСлучайный объект (первые 200 символов):")
print(f"'{random_chunk[:200]}...'")

# Токенизируем первые 30 слов из объекта
words_to_test = random_chunk.split()[:30]
all_tokens = []
for word in words_to_test:
    tokens = bpe_tokenize(word, merges)
    all_tokens.extend(tokens)

print(f"\nТокенизировано слов: {len(words_to_test)}")
print(f"Получено токенов: {len(all_tokens)}")
print(f"Первые 30 токенов: {all_tokens[:30]}")

# 10. Тест на незнакомом слове
test_word = "токенизацияпрекрасна"
print(f"\n--- Тест на незнакомом слове: '{test_word}' ---")
print(f"Токены: {bpe_tokenize(test_word, merges)}")
```

```python
"""
Пункт 3.3 для wikitext: Энтропия и дубликаты (самодостаточная ячейка)
"""
import numpy as np
import torch
import os
from collections import Counter
from tqdm import tqdm
from transformers import GPT2LMHeadModel, GPT2TokenizerFast

# ============================================================
# 1. ЯВНО ЗАГРУЖАЕМ GPT-2 (независимо от того, что было раньше)
# ============================================================
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Устройство: {device}")

print("Загрузка GPT-2...")
gpt2_tokenizer = GPT2TokenizerFast.from_pretrained("gpt2")
if gpt2_tokenizer.pad_token is None:
    gpt2_tokenizer.pad_token = gpt2_tokenizer.eos_token
gpt2_model = GPT2LMHeadModel.from_pretrained("gpt2").to(device)
gpt2_model.eval()
print("GPT-2 загружена.")

# ============================================================
# 2. ЗАГРУЖАЕМ ОЧИЩЕННЫЙ WIKITEXT
# ============================================================
input_path = "/content/drive/MyDrive/CommonCrawl/wikitext_cleaned.txt"
with open(input_path, 'r', encoding='utf-8') as f:
    text = f.read()
cleaned_wikitext = [c.strip() for c in text.split('\n\n') if c.strip()]
print(f"Загружено объектов: {len(cleaned_wikitext)}")

# ============================================================
# 3. ФУНКЦИЯ ПОДСЧЕТА ЭНТРОПИИ
# ============================================================
def calculate_entropy_batched(texts, batch_size=4):
    entropies = []
    for i in tqdm(range(0, len(texts), batch_size), desc="Энтропия"):
        batch_texts = texts[i:i+batch_size]
        encodings = gpt2_tokenizer(batch_texts, return_tensors="pt", padding=True,
                                    truncation=True, max_length=1024).to(device)
        with torch.no_grad():
            outputs = gpt2_model(**encodings)
            logits = outputs.logits
            shift_logits = logits[..., :-1, :].contiguous()
            shift_labels = encodings["input_ids"][..., 1:].contiguous()
            loss_fct = torch.nn.CrossEntropyLoss(reduction='none')
            losses = loss_fct(shift_logits.view(-1, shift_logits.size(-1)),
                              shift_labels.view(-1))
            losses = losses.view(shift_labels.size(0), -1)
            mask = encodings["attention_mask"][..., 1:].contiguous()
            for j in range(len(batch_texts)):
                valid_losses = losses[j][mask[j] == 1]
                entropies.append(valid_losses.sum().item())
    return entropies

# ============================================================
# 4. ПОДСЧЕТ ЭНТРОПИИ (подвыборка 500 объектов)
# ============================================================
sample_for_entropy = cleaned_wikitext
print(f"Подсчет энтропии для {len(sample_for_entropy)} объектов...")

entropies = calculate_entropy_batched(sample_for_entropy, batch_size=4)

# ============================================================
# 5. ФИЛЬТРАЦИЯ ПО ЭНТРОПИИ
# ============================================================
entropies_np = np.array(entropies)
low_threshold = np.percentile(entropies_np, 10)
high_threshold = np.percentile(entropies_np, 90)

print(f"\nПорог низкой энтропии (10%): {low_threshold:.2f}")
print(f"Порог высокой энтропии (90%): {high_threshold:.2f}")

filtered_wikitext = []
for chunk, ent in zip(sample_for_entropy, entropies):
    if low_threshold <= ent <= high_threshold:
        filtered_wikitext.append(chunk)

# ============================================================
# 6. ИНФОРМАЦИОННАЯ ПЛОТНОСТЬ
# ============================================================
total_entropy = sum(entropies)
total_tokens = sum([len(gpt2_tokenizer.encode(c)) for c in sample_for_entropy])
rho_info = total_entropy / total_tokens

print(f"\nИнформационная плотность (rho_info): {rho_info:.4f} бит/токен")
print(f"Объектов после фильтрации: {len(filtered_wikitext)}")

# ============================================================
# 7. СОХРАНЕНИЕ
# ============================================================
final_path = "/content/drive/MyDrive/CommonCrawl/wikitext_filtered.txt"
with open(final_path, 'w', encoding='utf-8') as f:
    f.write('\n\n'.join(filtered_wikitext))

print(f"\nФинальный датасет сохранен: {final_path}")
print(f"Размер: {os.path.getsize(final_path) / (1024*1024):.2f} MB")
```

```python
"""
Пункт 3.4 для wikitext: Единый формат вывода для трех токенизаторов
"""
import random
from collections import Counter

# 1. Загружаем отфильтрованный wikitext
input_path = "/content/drive/MyDrive/CommonCrawl/wikitext_filtered.txt"
with open(input_path, 'r', encoding='utf-8') as f:
    text = f.read()
wikitext_chunks = [c.strip() for c in text.split('\n\n') if c.strip()]

print(f"Загружено объектов wikitext: {len(wikitext_chunks)}\n")

# 2. Выбираем ОДИН случайный объект для всех трех токенизаторов
random.seed(42)  # Фиксируем seed для воспроизводимости
random_chunk = random.choice(wikitext_chunks)

print("=" * 80)
print("СЛУЧАЙНЫЙ ОБЪЕКТ (первые 200 символов):")
print("=" * 80)
print(f"'{random_chunk[:200]}...'")
print(f"Длина объекта: {len(random_chunk)} символов, {len(random_chunk.split())} слов\n")


# ============================================================
# 1. CHAR-LEVEL
# ============================================================
wiki_chars = set()
for chunk in wikitext_chunks:
    wiki_chars.update(chunk)

wiki_char_vocab = ['<PAD>', '<UNK>'] + sorted(list(wiki_chars))
wiki_char_to_idx = {c: i for i, c in enumerate(wiki_char_vocab)}
wiki_idx_to_char = {i: c for i, c in enumerate(wiki_char_vocab)}

char_tokens = [wiki_char_to_idx.get(c, wiki_char_to_idx['<UNK>']) for c in random_chunk]
char_decoded = ''.join([wiki_idx_to_char[i] for i in char_tokens])

print("=" * 80)
print("1. CHAR-LEVEL ТОКЕНИЗАТОР (обучен на wikitext)")
print("=" * 80)
print(f"Размер словаря:               {len(wiki_char_vocab)}")
print(f"Размер последовательности:     {len(char_tokens)}")
print(f"OOV (неизвестных символов):    {sum(1 for c in random_chunk if c not in wiki_char_to_idx)}")
print(f"Декодирование работает:        {char_decoded == random_chunk}")
print(f"Первые 30 токенов (ID):        {char_tokens[:30]}")
print(f"Первые 30 токенов (текст):     {[wiki_idx_to_char[i] for i in char_tokens[:30]]}")


# ============================================================
# 2. WORD-LEVEL
# ============================================================
all_words = []
for chunk in wikitext_chunks:
    all_words.extend(chunk.lower().split())

word_counts = Counter(all_words)
wiki_word_list = [w for w, c in word_counts.items() if c >= 2]
wiki_word_vocab = ['<PAD>', '<UNK>'] + sorted(wiki_word_list)
wiki_word_to_idx = {w: i for i, w in enumerate(wiki_word_vocab)}
wiki_idx_to_word = {i: w for i, w in enumerate(wiki_word_vocab)}

words = random_chunk.lower().split()
word_tokens = [wiki_word_to_idx.get(w, wiki_word_to_idx['<UNK>']) for w in words]

oov_count = sum(1 for w in words if w not in wiki_word_to_idx)
oov_ratio = oov_count / len(words) * 100

print("\n" + "=" * 80)
print("2. WORD-LEVEL ТОКЕНИЗАТОР (обучен на wikitext)")
print("=" * 80)
print(f"Размер словаря:               {len(wiki_word_vocab)}")
print(f"Размер последовательности:     {len(word_tokens)}")
print(f"OOV (неизвестных слов):        {oov_count} из {len(words)} ({oov_ratio:.2f}%)")
print(f"Первые 20 токенов (ID):        {word_tokens[:20]}")
print(f"Первые 20 токенов (текст):     {[wiki_idx_to_word[i] for i in word_tokens[:20]]}")


# ============================================================
# 3. BPE (используем готовый из Common Crawl)
# ============================================================
print("\n" + "=" * 80)
print("3. BPE-ТОКЕНИЗАТОР (обучен на Common Crawl, не переобучался)")
print("=" * 80)

try:
    _ = tokenizer.get_vocab_size()
    print("BPE tokenizer найден в памяти.")
except NameError:
    print("BPE tokenizer не найден. Обучаем заново на Common Crawl...")
    from tokenizers import Tokenizer
    from tokenizers.models import BPE
    from tokenizers.trainers import BpeTrainer
    from tokenizers.pre_tokenizers import Whitespace

    with open("/content/drive/MyDrive/CommonCrawl/filtered_dataset_v2.txt", 'r', encoding='utf-8') as f:
        cc_text = f.read()
    cc_chunks = [c.strip() for c in cc_text.split('\n\n') if c.strip()]

    tokenizer = Tokenizer(BPE(unk_token="<UNK>"))
    tokenizer.pre_tokenizer = Whitespace()
    trainer = BpeTrainer(vocab_size=10000, special_tokens=["<PAD>", "<UNK>", "<BOS>", "<EOS>"], min_frequency=2)
    tokenizer.train_from_iterator(cc_chunks, trainer=trainer)

encoding = tokenizer.encode(random_chunk)
oov_bpe = encoding.tokens.count('<UNK>')

print(f"Размер словаря:               {tokenizer.get_vocab_size()}")
print(f"Размер последовательности:     {len(encoding.ids)}")
print(f"OOV (неизвестных токенов):     {oov_bpe}")
print(f"Первые 30 токенов (ID):        {encoding.ids[:30]}")
print(f"Первые 30 токенов (текст):     {encoding.tokens[:30]}")


# ============================================================
# СВОДНАЯ ТАБЛИЦА
# ============================================================
print("\n" + "=" * 80)
print("СВОДНАЯ ТАБЛИЦА (wikitext)")
print("=" * 80)
print(f"{'Метод':<15} {'Размер словаря':<18} {'Длина последоват.':<20} {'OOV':<10}")
print("-" * 65)
print(f"{'Char-level':<15} {len(wiki_char_vocab):<18} {len(char_tokens):<20} {'0 (по опред.)':<10}")
print(f"{'Word-level':<15} {len(wiki_word_vocab):<18} {len(word_tokens):<20} {oov_ratio:.2f}%")
print(f"{'BPE (CC)':<15} {tokenizer.get_vocab_size():<18} {len(encoding.ids):<20} {oov_bpe}")
```

```python
"""
Пункт 3.5: Packed Batching
"""
import numpy as np
from collections import defaultdict

# 1. Токенизируем весь wikitext с помощью BPE (обученного на Common Crawl)
print("Токенизация wikitext через BPE...")

all_token_ids = []
for chunk in wikitext_chunks:
    encoding = tokenizer.encode(chunk)
    all_token_ids.append(encoding.ids)

total_tokens = sum(len(ids) for ids in all_token_ids)
print(f"Всего объектов: {len(all_token_ids)}")
print(f"Всего токенов: {total_tokens}")
print(f"Средняя длина объекта: {total_tokens / len(all_token_ids):.1f} токенов")
print(f"Минимальная длина: {min(len(ids) for ids in all_token_ids)}")
print(f"Максимальная длина: {max(len(ids) for ids in all_token_ids)}")


# ============================================================
# 2. PACKED BATCHING
# ============================================================
def packed_batching(all_token_ids, batch_size, pad_token_id=0):
    """
    Упаковывает объекты в батчи фиксированной длины.

    Возвращает:
    - packed_batches: список батчей, каждый - список токенов длины batch_size
    - attention_masks: маски для каждого батча (0 = pad, 1, 2, 3... = номер объекта)
    """
    packed_batches = []
    attention_masks = []

    # Текущий батч и его маска
    current_batch = []
    current_mask = []
    object_counter = 1  # Номер объекта внутри текущего батча

    for token_ids in all_token_ids:
        i = 0
        while i < len(token_ids):
            # Сколько токенов можем добавить в текущий батч
            space_left = batch_size - len(current_batch)
            tokens_to_add = token_ids[i:i + space_left]

            # Добавляем токены и маску (номер объекта)
            current_batch.extend(tokens_to_add)
            current_mask.extend([object_counter] * len(tokens_to_add))

            i += space_left
            object_counter += 1  # Начинаем новый объект

            # Если батч заполнен — сохраняем и начинаем новый
            if len(current_batch) == batch_size:
                packed_batches.append(current_batch)
                attention_masks.append(current_mask)
                current_batch = []
                current_mask = []
                object_counter = 1

    # Обрабатываем последний неполный батч (добиваем PAD)
    if current_batch:
        pad_count = batch_size - len(current_batch)
        current_batch.extend([pad_token_id] * pad_count)
        current_mask.extend([0] * pad_count)  # 0 = PAD
        packed_batches.append(current_batch)
        attention_masks.append(current_mask)

    return packed_batches, attention_masks


# ============================================================
# 3. ПРИМЕНЯЕМ PACKED BATCHING
# ============================================================
BATCH_SIZE = 512  # Длина одной последовательности (из задания: 512-1024)

print(f"\nПрименяем packed batching (batch_size={BATCH_SIZE})...")
packed_batches, attention_masks = packed_batching(all_token_ids, batch_size=BATCH_SIZE)

print(f"\n--- Результаты Packed Batching ---")
print(f"Количество батчей: {len(packed_batches)}")
print(f"Общая длина всех батчей: {sum(len(b) for b in packed_batches)} токенов")
print(f"Полезных токенов: {total_tokens}")
print(f"PAD-токенов: {sum(len(b) for b in packed_batches) - total_tokens}")
print(f"Эффективность: {total_tokens / sum(len(b) for b in packed_batches) * 100:.2f}%")


# ============================================================
# 4. СРАВНЕНИЕ С PADDING (без packing)
# ============================================================
max_len = max(len(ids) for ids in all_token_ids)
padded_total = len(all_token_ids) * max_len

print(f"\n--- Сравнение с обычным паддингом ---")
print(f"Обычный паддинг (все до {max_len}):")
print(f"  Общая длина: {padded_total} токенов")
print(f"  Эффективность: {total_tokens / padded_total * 100:.2f}%")
print(f"Packed batching (batch_size={BATCH_SIZE}):")
print(f"  Общая длина: {sum(len(b) for b in packed_batches)} токенов")
print(f"  Эффективность: {total_tokens / sum(len(b) for b in packed_batches) * 100:.2f}%")


# ============================================================
# 5. ДЕМОНСТРАЦИЯ
# ============================================================
print(f"\n--- Пример первого батча ---")
print(f"Токены (первые 50):     {packed_batches[0][:50]}")
print(f"Маска внимания (первые 50): {attention_masks[0][:50]}")

# Покажем, где начинаются новые объекты (изменение маски)
print(f"\nГраницы объектов в первом батче:")
prev_mask = 0
for i, m in enumerate(attention_masks[0]):
    if m != prev_mask and m != 0:
        print(f"  Позиция {i}: начало объекта #{m}")
    prev_mask = m

print(f"\n--- Пример последнего батча (с PAD) ---")
print(f"Токены (последние 50):      {packed_batches[-1][-50:]}")
print(f"Маска внимания (последние 50): {attention_masks[-1][-50:]}")
```
